                          Data cleaning in Pandas

```python
df9 = pd.read_excel(r"C:\Users\ymogo\OneDrive\Desktop\Data_analytic\pandas\PandasYouTubeSeries-main\PandasYouTubeSeries-main\Customer Call List.xlsx")
df9
```

```python
df9 = df9.drop_duplicates()
df9

```

```python
df9 = df9.drop(columns = 'Not_Useful_Column')
```

```python
df9
```

```python
#df9['Last_Name'] = df9['Last_Name'].str.lstrip('...')
#df9['Last_Name'] = df9['Last_Name'].str.lstrip('/')
#df9['Last_Name'] = df9['Last_Name'].str.rstrip('_')
df9['Last_Name'] = df9['Last_Name'].str.strip('123._/') # another way doing all at once
df9
```

```python
df9["Phone_Number"] = df9["Phone_Number"].str.replace('[^a-zA-Z0-9]','',regex=True)
df9
```

```python
#df9["Phone_Number"] = df9["Phone_Number"].apply(lambda x: str(x))
df9["Phone_Number"] = df9["Phone_Number"].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10])
df9
```

```python
df9["Phone_Number"] = df9["Phone_Number"].str.replace('nan--','')
df9["Phone_Number"] = df9["Phone_Number"].str.replace('Na--','')
df9
```

```python
df9[["Street_Address","State","Zip_Code"]] = df9["Address"].str.split(',',n=2,expand=True)
df9
```

```python
df9["Paying Customer"] = df9["Paying Customer"].str.replace("Yes","Y")
df9["Paying Customer"] = df9["Paying Customer"].str.replace("No","N")
df9
```

```python
df9["Do_Not_Contact"] = df9["Do_Not_Contact"].str.replace("Yes","Y")
df9["Do_Not_Contact"] = df9["Do_Not_Contact"].str.replace("No","N")
df9
```

```python
df9 = df9.replace("N/a","")
df9 = df9.fillna("")
df9
```

```python
for x in df9.index:
    if(df9.loc[x,"Do_Not_Contact"]) == "Y":
       df9.drop(x,inplace=True)
df9
 
```

```python
for x in df9.index:
    if(df9.loc[x,"Phone_Number"]) == "":
       df9.drop(x,inplace=True)
df9
# another way of dropping null value in a column
# df9 = df9.dropna(subset="Phone_Number"), inplace=True
```

```python
df9.drop(columns="Address")
```

```python
df9.reset_index(drop=True,inplace=True)
```
